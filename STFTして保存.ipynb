STFTの結果をUMAPに適用するために特定の周波数帯のデータセットをつくるプログラムの作成

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import glob
from scipy import signal
%matplotlib inline
```

```python
min=0
max=8000
N=max-min
sf=4000
nperseg=256
noverlap=128

Sxx1_all=[]
Sxx2_all=[]
Sxx3_all=[]

num=10
cou=0

name=['aa','ii','uu','ee','oo']

#file 読み込み
for name_num in range(len(name)):
  for file in glob.glob('drive/MyDrive/aaiiuueeoo/'+name[name_num]+'/*.csv'):
    print(file)
    cou=cou+1
    df=pd.read_csv(file,skiprows=13,header=None)
    data1=df[min:max][0]
    data2=df[min:max][1]
    data3=df[min:max][2]
    f1,t1,Sxx1=signal.spectrogram(data1,sf,nperseg=nperseg,noverlap=noverlap,window='hann')
    f2,t2,Sxx2=signal.spectrogram(data2,sf,nperseg=nperseg,noverlap=noverlap,window='hann')
    f3,t3,Sxx3=signal.spectrogram(data3,sf,nperseg=nperseg,noverlap=noverlap,window='hann')

    Sxx1_all=np.append(Sxx1_all,Sxx1[num],axis=0)
    Sxx2_all=np.append(Sxx2_all,Sxx2[num],axis=0)
    Sxx3_all=np.append(Sxx3_all,Sxx3[num],axis=0)

final1=Sxx1_all.reshape(cou,len(t1))
final2=Sxx2_all.reshape(cou,len(t1))
final3=Sxx3_all.reshape(cou,len(t1))

# final=np.append(final1,final2,axis=0)
# final=np.append(final,final3,axis=0)

print(final1.shape)
```

```python
np.save('drive/MyDrive/aaiiuueeoo/test1',final1)
np.save('drive/MyDrive/aaiiuueeoo/test2',final2)
np.save('drive/MyDrive/aaiiuueeoo/test3',final3)
```
