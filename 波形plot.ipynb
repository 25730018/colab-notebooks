# 母音単音

```python
path='drive/MyDrive/suzuki_Data/suzuki/emg2026-05-14-19-56-42/'
#name_input=['a','i','u','e','o']
name_input=['mosimosi1']
#name_input=['ariga-a','gomen-a','hazim-a','konnb-a','konnn-a','ohayo-a','oisii-a','onega-a','otuka-a','sayou-a','ariga-y','gomen-y','hazim-y','konnb-y','konnn-y','ohayo-y','oisii-y','onega-y','otuka-y','sayou-y']
min_t=0
max_t=6000
N=max_t-min_t
sf=4000

x = np.arange(N) / sf * 1000  # 単位: ms


for name in name_input:
    cou=0
    for file in glob.glob(path+name+'/*.csv'):
        cou=cou+1
        df=pd.read_csv(file,skiprows=13,header=None)
        data1=df[0:N][0]
        data2=df[0:N][1]
        data3=df[0:N][2]

        print(file)
        fig,axes=plt.subplots(3,1,tight_layout=True,sharex=True,sharey=True)
        axes[0].plot(x,data1,color='r',label='ch1')
        axes[0].set_ylim(-0.1,0.1)
        axes[1].plot(x,data2,color='b',label='ch2')
        axes[2].plot(x,data3,color='g',label='ch3')
        axes[0].legend(loc='upper right')
        axes[1].legend(loc='upper right')
        axes[2].legend(loc='upper right')
        axes[1].set_ylabel('EMG [V]')
        axes[2].set_xlabel('Time [ms]')
        plt.show()
        plt.clf()
        plt.close()
        # print('************')
        if cou==5:
            break
```

# コマンド

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import glob
import os

# 対象のカテゴリ名を指定（例：'hutuu'）
name = 'teisoku'

# パス設定
path = 'drive/MyDrive/suzuki_Data/suzuki/2025-07-17/emg2025-07-17-15-53-59/'

N = 9313
sf = 4000
x = np.arange(0, N/sf, 1/sf)  # 時間軸（秒）

cou = 0

# ファイル検索
csv_files = glob.glob(os.path.join(path, name, '*.csv'))

for file in csv_files:
    cou += 1
    df = pd.read_csv(file, skiprows=13, header=None)

    # 長さ確認（足りない場合スキップ）
    if len(df) < N:
        print(f"スキップ: {file} （データ長不足: {len(df)} < {N}）")
        continue

    data1 = df.iloc[0:N, 0]
    data2 = df.iloc[0:N, 1]
    data3 = df.iloc[0:N, 2]

    print(f"表示中: {file}")
    fig, axes = plt.subplots(3, 1, tight_layout=True, sharex=True, sharey=True)

    axes[0].plot(x, data1, color='r', label='ch1')
    axes[0].set_ylim(-0.1, 0.1)

    axes[1].plot(x, data2, color='b', label='ch2')
    axes[2].plot(x, data3, color='g', label='ch3')

    for ax in axes:
        ax.legend(loc='upper right')

    axes[1].set_ylabel('EMG [V]')
    axes[2].set_xlabel('Time [s]')

    plt.show()
    plt.clf()
    plt.close()

    if cou == 5:
        break
```
